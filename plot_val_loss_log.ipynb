```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# val_loss_log.npy 파일 경로 지정 및 데이터 로드
val_loss_log_path = Path('../result/test_Unet/val_loss_log.npy')  # 필요시 경로 수정
val_loss_log = np.load(val_loss_log_path)
print(f"val_loss_log shape: {val_loss_log.shape}")
print(val_loss_log[:5])  # 앞부분 미리보기
```

```python
# val_loss_log 그래프 그리기
epochs = val_loss_log[:, 0]
val_losses = val_loss_log[:, 1]

plt.figure(figsize=(8,5))
plt.plot(epochs, val_losses, marker='o', label='Val Loss')
plt.xlabel('Epoch')
plt.ylabel('Validation Loss')
plt.title('Validation Loss per Epoch')
plt.grid(True)
plt.legend()
plt.show()
```
